```python
customers=pd.read_csv("customers.csv")
order_items=pd.read_csv("order_items.csv")
orders=pd.read_csv("orders.csv")
payments=pd.read_csv("payments.csv")
reviews=pd.read_csv("reviews.csv")
products=pd.read_csv("products.csv")
```

```python
#merge customers.csv and orders.csv using customer_id and display customer_id, custimer name,order_id and order_date
merge_res=pd.merge(customers,orders,left_on="customer_id",right_on="customer_id",how="inner")
merge_res
```

```python
#Merge order.csv and payments.csv using order_id and display order details along with payment method, payment amount and payment status
new1=payments[["order_id","payment_method","amount","payment_status"]]
res=pd.merge(orders,new1,left_on="order_id",right_on="order_id",how="inner")
res
```

```python
#merge customers.csv and payments.csv and display customer name, order_id, payment method and payment amount
new1=customers[["customer_id","name"]]
new2=orders[["customer_id","order_id"]]
new3=payments[["order_id","payment_method","amount"]]
res=pd.merge(new1,new2,left_on="customer_id",right_on="customer_id",how="inner")
result=pd.merge(res,new3,left_on="order_id",right_on="order_id",how="inner")
result
```

```python
#merge order_items.csv and products.csv using product id and display order_id, product id, product_name, categort, quamtity and total_price.
new1=order_items[["product_id","order_id","total_price"]]
new2=products[["product_id","product_name","category",	"stock"]]
res=pd.merge(new1,new2,left_on="product_id",right_on="product_id",how="inner")
res
```

```python
#merge order_items.csv with products.csv and find the total revenue generated by each product
new1=order_items[["product_id","total_price"]]
new2=products[["product_id","product_name"]]
res=pd.merge(new1,new2,left_on="product_id",right_on="product_id",how="inner")
res.groupby(["product_name"])["total_price"].sum().reset_index()
```

```python
#merge orders.csv,  order_items.csv and customers.csv and find the total spending of each customer. 
new1=orders[["customer_id", "order_id"]]
new2=order_items[["order_id","total_price"]]
new3=customers[["customer_id","name"]]

res1=pd.merge(orders,order_items,left_on="order_id",right_on="order_id",how="inner")
res2=pd.merge(res1,customers,left_on="customer_id",right_on="customer_id",how="inner")
res2.groupby(["name"])["total_price"].sum().reset_index()
```

```python
#using the customer spending data, find the top 10 customers based on total spending
new1=orders[["customer_id", "order_id"]]
new2=order_items[["order_id","total_price"]]
new3=customers[["customer_id","name"]]

res1=pd.merge(orders,order_items,left_on="order_id",right_on="order_id",how="inner")
res2=pd.merge(res1,customers,left_on="customer_id",right_on="customer_id",how="inner")
res2.groupby(["name"])["total_price"].sum().sort_values(ascending=False).head(10)
```

```python
#merge order_items.csv with products.csv and find the total revenue generated by each product category
new1=order_items[["product_id","total_price"]]
new2=products[["product_id","category"]]
res=pd.merge(new1,new2,left_on="product_id",right_on="product_id",how="inner")
res.groupby(["category"])["total_price"].sum()
```

```python
#Merge customers.csv, orders.csv, and order_items.csv and 
#find the total revenue generated by each city.
new1=customers[["customer_id","city"]]
new2=orders[["customer_id","order_id"]]
new3=order_items[["order_id","total_price"]]
res1=pd.merge(new1,new2,left_on="customer_id",right_on="customer_id",how="inner")
result=pd.merge(res1,new3,left_on="order_id",right_on="order_id",how="inner")

result.groupby(["city"])["total_price"].sum().reset_index()
```

```python
#Merge reviews.csv with products.csv and find the average rating for every product
new1=reviews[["product_id","rating"]]
new2=products[["product_id","product_name"]]
res=pd.merge(new1,new2,left_on="product_id",right_on="product_id",how="inner")
res.groupby(["product_name"])["rating"].mean().reset_index()
```

```python
#merge products.csv , order_items.csv and reviews.csv and create a dataframe containing product name, category, total_quantity sold, 
#total revenue, average rating
new1=products[["product_id","product_name","category"]]
new2=order_items[["product_id","quantity","total_price"]]
new3=reviews[["product_id","rating"]]

res1=pd.merge(new1,new2,left_on="product_id",right_on="product_id",how="inner")
res2=pd.merge(res1,new3,left_on="product_id",right_on="product_id",how="inner")

res2.groupby(["product_id","product_name","category"]).agg(total_quantity=("quantity","sum"),
                                                        total_revenue=("total_price","sum"),
                                                        average_rating=("rating","mean")).reset_index()
```
