```python
#Retrieve students who have GRE Score divisible by 5.  Add a new
#column named GRE_to_TOEFL_Ratio = GRE Score / TOEFL Score.
gre=df[df["GRE Score"]%5==0]
df["GRE_to_TOEFL_RATIO"]=df["GRE Score"]/df["TOEFL Score"]
df["GRE_to_TOEFL_RATIO"]
```

```python
#Create a new column Admission_Chance_% = Chance of Admit × 100.
df["Admission_Chance_%"]=df["Chance of Admit"]*100
df["Admission_Chance_%"]
```

```python
#Rename the column Chance of Admit to Admit_Probability.
df=df.rename(columns={"Chance of Admit":"Admit_Probability"})
df.columns
```

```python
#Change all column names to lowercase.
df.columns = df.columns.str.lower()
```

```python
df.columns
```

```python
#Drop the column Serial No..
df.drop(columns=["serial no."],inplace=True)
```

```python
#Round all values in CGPA to one decimal place.
df["cgpa"] = df["cgpa"].round(1)
```

```python
#Replace all University Rating values of 5 with 10.
df["university rating"]=df["university rating"].replace(5,10)
```

```python
df["university rating"].unique()
```

```python
#Convert Research column values from 0/1 to “No”/“Yes”.
df["research"]=df["research"].apply(lambda x: "Yes" if x==1 else "No")
```

```python
#Create a new column GRE_Level = “High” if GRE ≥ 320 else “Low”.
df["GRE_Level"]=df["gre score"].apply(lambda x: "High" if x>=320 else "Low")
```

```python
#Normalize CGPA column to range 0–1 using min-max scaling (without
#sklearn).
df["cgpa_normalized"]=(df["cgpa"]-df["cgpa"].min())/(df["cgpa"].max()-df["cgpa"].min())
```

```python
#Increase all TOEFL Score values by 2 points.
df["toefl score"]=df["toefl score"]+2
```

```python
#Add a column Weighted_Score = GRE Score * 0.3 + TOEFL Score * 0.2 +
#CGPA * 0.5.
df["weighted score"]=(df["gre score"]*0.3)+(df["toefl score"]*0.2)+(df["cgpa"]*0.5)
```

```python
df=df.rename(columns={"GRE_Level":"gre level"})
```

```python
df.columns
```

```python
#Replace all SOP values less than 2 with the mean SOP.
df.loc[df["sop"]<2,"sop"]=df["sop"].mean()
#loc-select a particular row or column
```

```python
#Create a new column combining University Rating and Research (e.g.,
#Rating-Research).
df["rating research"] = df["university rating"].astype(str) + "-" + df["research"].astype(str)
```

```python
#Convert Chance of Admit to categorical bins: “Low”, “Medium”, “High”.
df["admission_chance_%"]=df["admission_chance_%"].apply(bins)
```

```python
def bins(x):
    if x==100:
        return "High"
    elif x==0:
        return "Low"
    else:
        return "Medium"
```

```python
df["admission_chance_%"].unique()
```

```python
#Add a constant column Country with value “India”.
df["country"] = "India"
```

```python
#Reorder columns so that Chance of Admit comes first.
df = df[["admission_chance_%"] + [col for col in df.columns if col != "admission_chance_%"]]
```

```python
#Replace values in LOR column where rating is 5 with 4.9.
df["lor"]=df["lor"].replace(5,4.9)
```

```python
df["lor"].unique()
```

```python
df.info()
```

```python
##Convert all float columns to integers (after rounding).
df["sop"]=df["sop"].astype(int)
df["lor"]=df["lor"].astype(int)
df["cgpa"]=df["cgpa"].astype(int)
df["gre_to_toefl_ratio"]=df["gre_to_toefl_ratio"].astype(int)
df["cgpa_normalized"]=df["cgpa_normalized"].astype(int)
df["weighted score"]=df["weighted score"].astype(int)
```

```python
df.info()
```

```python
#Replace CGPA values above 9.5 with 9.5.
df["cgpa"] = df["cgpa"].astype(float)
df.loc[df["cgpa"]>9.5,"cgpa"]=9.5
```

```python
#Apply a lambda function to square all SOP values.
df["sop"]=df["sop"].apply(lambda x: x**2)
```

```python
#Add a column GRE_Category = “Below Avg”, “Avg”, “Above Avg” based
#on mean GRE.
mean_gre = df["gre score"].mean()

def gre_category(x):
    if x < mean_gre:
        return "Below Avg"
    elif x > mean_gre:
        return "Above Avg"
    else:
        return "Avg"

df["gre category"] = df["gre score"].apply(gre_category)
```
