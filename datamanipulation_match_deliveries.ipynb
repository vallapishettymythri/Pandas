```python
#Join matches and deliveries to find total balls bowled per match.
res=pd.merge(df1,df2,left_on="id",right_on="match_id",how="inner")
total_balls=res.groupby("id")["ball"].count().reset_index()
total_balls
```

```python
#For each match, get total runs scored by joining matches and deliveries.
new1=df1[["id"]]
new2=df2[["match_id","total_runs"]]
res=pd.merge(new1,new2,left_on="id",right_on="match_id",how="inner")
res.groupby("id")["total_runs"].sum().reset_index()
```

```python
#Find the top 10 highest-scoring matches.
res=pd.merge(df1,df2,left_on="id",right_on="match_id",how="inner")
res.groupby(["id"])["total_runs"].sum().sort_values(ascending=False).head(10)
```

```python
#Find the number of wickets per match using join.
res=pd.merge(df1,df2,left_on="id",right_on="match_id",how="inner")
wickets=res[res["is_wicket"]==1]
wickets.groupby(["id"])["is_wicket"].count()
```

```python
#List matches where more than 15 wides were bowled.
#df2-extras_type-wides
res=pd.merge(df1,df2,left_on="id",right_on="match_id",how="inner")
wide=res[res["extras_type"]=="wides"]
wide_count=wide.groupby(["id"])["extras_type"].count()
wide_count[wide_count>15].reset_index()
```

```python
#Find the player with the most deliveries faced across all matches.
deliveries_count=df2.groupby("batter")["ball"].count()
deliveries_count.sort_values(ascending=False).head(1)
```

```python
#Find total boundaries (4s &amp; 6s) per match by joining datasets.
def boundaries(df1,df2):
    res=pd.merge(df1,df2,left_on="id",right_on="match_id",how="inner")

    boundaries=res[res["batsman_runs"]==4 | (res["batsman_runs"]==6)]
    result=boundaries.groupby("id")["batsman_runs"].count()
    return result
boundaries(df1,df2)
```

```python
#Combine matches and deliveries to compute strike rate of each batsman
#in each match.
#Strike Rate = (Runs scored / Balls faced) × 100
res=pd.merge(df1,df2,left_on="id",right_on="match_id",how="inner")
result=res.groupby(["id","batter"]).agg({"batsman_runs":"sum","ball":"count"})
result["strike_rate"]=(result["batsman_runs"]/result["ball"])*100
result
```

```python
#For each season, get total runs scored (join matches → deliveries).
res=pd.merge(df1,df2,left_on="id",right_on="match_id",how="inner")
res.groupby(["season"])["total_runs"].sum().reset_index()
```

```python
#For each season, find total wickets taken.
res=pd.merge(df1,df2,left_on="id",right_on="match_id",how="inner")
wickets=res[res["is_wicket"]==1]
wickets.groupby(["season"])["is_wicket"].sum().reset_index()
```

```python
#Find bowlers who conceded more than 50 runs in a match.
res = pd.merge(df1, df2, left_on="id", right_on="match_id", how="inner")

runs = res.groupby(["id", "bowler"])["total_runs"].sum()

runs[runs > 50]
```

```python
#For each match, find the batsman with the highest strike rate (min 10
#balls).
result = df2.groupby(["match_id", "batter"]).agg({
    "batsman_runs": "sum",
    "ball": "count"
})

result = result[result["ball"] >= 10]

result["strike_rate"] = (result["batsman_runs"] / result["ball"]) * 100

result
```

```python
#Calculate the economy of each bowler per season (join deliveries →
#matches).
result=pd.merge(df1,df2,left_on="id",right_on="match_id",how="inner")
result["bowler_runs"]=result["total_runs"]-result["batsman_runs"]
res = result.groupby(["season", "bowler"]).agg({
    "bowler_runs": "sum",
    "ball": "count"
})
res["economy"] = (res["bowler_runs"] / res["ball"]) * 6
res
```
