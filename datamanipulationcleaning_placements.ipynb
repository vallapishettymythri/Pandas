```python
#Add a new column overall_avg = (ssc_p + hsc_p + degree_p + mba_p) / 4.
df["overall_avg"]=((df["ssc_p"]+df["hsc_p"]+df["degree_p"]+df["mba_p"])/4)
```

```python
df["overall_avg"]
```

```python
#Rename the column degree_p to degree_percentage.
df=df.rename(columns={"degree_p":"degree_per"})
```

```python
#Create a new column placement_status = “Hired” if status = “Placed” else “Not
#Hired”.
df["placement_status"] = df["status"].apply(
    lambda x: "Hired" if x == "Placed" else "Not Hired"
)
```

```python
df["placement_status"].unique()
```

```python
df["status"].unique()
```

```python
#Drop the column sl_no.
df.drop(columns="sl_no",inplace=True)
```

```python
#Replace all “Comm&amp;Mgmt” in degree_t with “Commerce”.
df["degree_t"]=df["degree_t"].replace("Comm&Mgmt","Commerce")
```

```python
df["degree_t"].unique()
```

```python
#Convert gender column values “M” → “Male”, “F” → “Female”.
df["gender"]=df["gender"].apply(lambda x: "Male" if x=="M" else "Female")
```

```python
#Create a new column above_80_ssc = True if ssc_p &gt; 80 else False.
df["above_80_ssc"]=df["ssc_p"].apply(lambda x: "True" if x>80 else "False")
```

```python
#Round all percentage columns to 1 decimal place.
df["ssc_p"]=df["ssc_p"].round(1)
df["hsc_p"]=df["hsc_p"].round(1)
df["degree_per"]=df["degree_per"].round(1)
df["etest_p"]=df["etest_p"].round(1)
df["mba_p"]=df["mba_p"].round(1)
df["salary"]=df["salary"].round(1)
```

```python
df.info()
```

```python
#Convert all column names to lowercase.
df.columns = df.columns.str.lower()
```

```python
#Add a new column salary_in_lpa = salary / 100000.
df["salary_in_lpa"]=(df["salary"]/100000)
```

```python
#Replace NaN values in salary with 0.
df["salary"] = df["salary"].fillna(0)
```

```python
#Drop all duplicate rows from the dataset.
df.duplicated().sum()
```

```python
#Increase all mba_p scores by 5 points.
df["mba_p"]=(df["mba_p"]+5)
```

```python
#Convert salary column to integer type.
df["salary"]=df["salary"].astype(int)
```

```python
df.info()
```

```python
#Reorder columns to make status the first column.
df = df[["status"] + [col for col in df.columns if col != "status"]]
```

```python
#Replace all “Yes”/“No” in workex with 1/0.
df["workex"]=df["workex"].apply(lambda x: 1 if x=="Yes" else 0)
```

```python
df["workex"].unique()
```

```python
#Create a new column performance_tag: “Excellent” if degree_p &gt; 80 else
#“Average”.
df["performance_tag"]=df["degree_per"].apply(lambda x: "Excellent" if x>80 else "Average")
```

```python
#Standardize all column names to snake_case (e.g., ssc_p → ssc_percentage).
df = df.rename(columns={
    "ssc_p": "ssc_percentage",
    "hsc_p": "hsc_percentage",
    "degree_per": "degree_percentage",
    "etest_p": "etest_percentage",
    "mba_p": "mba_percentage"
})
```

```python
df.columns
```

```python
#Create a new column combining degree_t and specialisation (e.g.,
#“Sci&amp;Tech_Mkt&amp;HR”).
df["degree_specialisation"] = df["degree_t"] + "_" + df["specialisation"]
```

```python
#Fill missing etest_p values with mean of that column.
mean=df["etest_percentage"].mean()
df["etest_percentage"]=df["etest_percentage"].fillna(mean)
```

```python
#Create a new column total_score = ssc_p + hsc_p + degree_p + mba_p.
df["total_score"]=df["ssc_percentage"]+df["hsc_percentage"]+df["degree_percentage"]+df["mba_percentage"]
```

```python
#Create a boolean column is_placed = True if status = “Placed”.
df["is_placed"]=df["status"].apply(lambda x: True if x=="Placed" else False)
```

```python
#Add a column mba_grade = “A” if mba_p ≥ 75 else “B”.
df["mba_grade"]=df["mba_percentage"].apply(lambda x: "A" if x>=75 else "B")
```

```python
#Replace NaN salaries for non-placed students with 0.
df.loc[df["status"] == "Not Placed", "salary"] = 0
```
